# StandardScaler in Scikit-Learn.

```python
# StandardScaler fixes this. It looks at Class A, sees the 90 average, and realizes a 70 is actually quite poor (a negative score). It looks at Class B, sees the 40 average, and realizes a 70 is actually amazing (a high positive score). It turns every grade into a "distance from the average" so you can compare students from totally different classes fairly.
```

```python
from sklearn.preprocessing import StandardScaler

data = np.array([[14, 28], [32, 49], [53, 68], [93, 46], [58, 60]])

scaler = StandardScaler()

scaler_data = scaler.fit_transform(data)

df = pd.DataFrame(scaler_data)

print(df)
```
